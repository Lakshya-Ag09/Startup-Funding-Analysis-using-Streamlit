```python
def startup_name_cleaning(text):
    if ('.in' in text) | ('.co' in text):
            if '.in' in text:
                    idx_end = text.index('.in')
            if '.co' in text:
                    idx_end = text.index('.co')
            
            if 'www.' in text:
                    idx_start = text.index('www.')
                    text = text[idx_start+4:idx_end]
            else:
                    text = text[:idx_end]
            return text
    elif '\\' in text:
            idx = text.index('\\')
            if text[idx+2] != 'x':
                    text = text.replace('\\','')
                    return text
            else:
                    for i in range(idx+3,len(text)):
                            if text[i] == 's':
                                    break
                    text = text.replace(text[idx:i+1], '')
                    return text
    else:
            return text
```

```python
df['Startup Name'] = df['Startup Name'].apply(lambda x: startup_name_cleaning(x))
```

```python
def to_inr(dollar):
    inr = dollar*95
    return inr/10000000
```
